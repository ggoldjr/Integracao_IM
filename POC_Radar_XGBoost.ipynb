```python
import pandas as pd
import joblib
```

```python
#Dados
df_test = pd.read_csv("df_test.csv")
```

```python
## Modelo de regressão
MODEL_FEATURES= ["rpm", "wind_speed", "current", "temperature", "vibration", "time_days"]
X_test = df_test[MODEL_FEATURES]
print("\nDados da observação:")

# Carregar o modelo 
modelo_xgb_reg = joblib.load("model_xgb_reg.pkl")

# teste
i = [10, 50]  # indice da tabela
X_observacao = X_test.iloc[i]
display(X_observacao)

# Predição utilizando o modelo carregado do arquivo .pkl
pred_xgb_reg = modelo_xgb_reg.predict(X_observacao)

# Criar a tabela de resultados
resultados = df_test.iloc[i].copy()

resultados["Predição RUL - XGBoost (.pkl)"] = pred_xgb_reg.round(2)

print("\nResultado real e predição:")
display(resultados)
```

```python
## Modelo de classificação
# Carregar o modelo salvo
modelo_xgb_cls = joblib.load("model_xgb_cls.pkl")
label_encoder = joblib.load("label_encoder.pkl")


# Predição utilizando o modelo carregado do arquivo .pkl
predicao_codificada = modelo_xgb_cls.predict(X_observacao).astype(int)
pred_xgb_cls = label_encoder.inverse_transform(predicao_codificada)

# Criar a tabela de resultados
resultados = df_test.iloc[i].copy()

resultados["Predição RUL - XGBoost (.pkl)"] = pred_xgb_cls

print("\nResultado real e predição:")
display(resultados)
```
